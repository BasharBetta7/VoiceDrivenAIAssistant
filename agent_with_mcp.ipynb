```python
import asyncio

from langchain.agents import create_agent
from langchain.mcp import MCPAdapter
from langchain_ollama import ChatOllama


async def main():
    config = {
        "mcpServers": {
            "weather": {
                "command": "npx",
                "args": ["-y", "@dangahagan/weather-mcp"],
                "env": {"WEATHER_UNITS": "metric"},
            },
            "sequential-thinking": {
                  "command": "npx",
                  "args": [
                    "-y",
                    "@modelcontextprotocol/server-sequential-thinking"
                  ]
            },
        }
    }


    async with MCPAdapter(config) as adapter:
        weather_tools = await adapter.list_tools()

        print("Available tools:", [t.name for t in weather_tools])

        agent = create_agent(
            model=ChatOllama(
                model="qwen3.5:4b",
                temperature=0,
            ),
            tools=weather_tools,
        )

        result = await agent.ainvoke({
            "messages": [{
                "role": "user",
                "content": "list available lessons",
            }]
        })

        print(result["messages"][-1].content)


await main()
```
