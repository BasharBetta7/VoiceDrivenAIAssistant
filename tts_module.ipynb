```python
import torch
import sounddevice as sd
from qwen_tts import Qwen3TTSModel

# Load the model
model = Qwen3TTSModel.from_pretrained(
    "Qwen/Qwen3-TTS-12Hz-0.6B-Base",
    device_map="cuda:0",
    dtype=torch.bfloat16,
    attn_implementation="sdpa",
    
)


```

```python
import sounddevice as sd

audio = (sd.rec(
    samplerate=16000, frames= int(3 * 16000),channels=1,dtype="int16",
), 16000)
sd.wait()

import soundfile as sf

sf.write('audio_wavs/ref_audio.wav', audio[0], 16000)
```

```python
# Reference audio for cloning
ref_audio = "audio_wavs/tts_ref_audio.wav"
ref_text  = "Hello, this is a simple test."

voice_clone_prompt = model.create_voice_clone_prompt(
    ref_audio=ref_audio,
    ref_text=ref_text
)
# Generate speech
wavs, sr = model.generate_voice_clone(
    text="Yeah sure, you are right, but In my opinion, being right is not enough.",
    language="English",
    voice_clone_prompt=voice_clone_prompt,
    pad_token_id=2150
    
)

# Play the resulting audio directly from memory
sd.play(wavs[0], samplerate=sr)
sd.wait()
```
